# DATA Cleaning

```python
data = raw_df.iloc[3:].copy()

data = data[
    ~data[0].isin([
        "جمع به تفکیک هر واحد",
        "کل شرکت"
    ])
].copy()

data.reset_index(drop=True, inplace=True)

print(data.shape)
data.head()
```

```python
data.shape
```

```python
data = data.rename(columns={
    0: "city_name",
    1: "city_code",
    2: "year",
    3: "month"
})
```

```python
data["city_name"] = data["city_name"].astype(str).str.strip()

data["city_code"] = pd.to_numeric(
    data["city_code"],
    errors="raise"
).astype(int)

data["year"] = pd.to_numeric(
    data["year"],
    errors="raise"
).astype(int)

data["month"] = pd.to_numeric(
    data["month"],
    errors="raise"
).astype(int)
```

```python
data[
    ["city_name", "city_code", "year", "month"]
].head()
```

```python
activity_groups = [
    {
        "activity": "test_1",
        "columns": {
            4: "in_progress",
            5: "statement_preparation",
            6: "at_headquarters",
            7: "at_finance"
        },
        "total_column": 8
    },
    {
        "activity": "test_2",
        "columns": {
            9: "in_progress",
            10: "statement_preparation",
            11: "at_headquarters",
            12: "at_finance"
        },
        "total_column": 13
    },
    {
        "activity": "test_3",
        "columns": {
            14: "in_progress",
            15: "statement_preparation",
            16: "at_headquarters",
            17: "at_finance"
        },
        "total_column": 18
    },
    {
        "activity": "test_4_a",
        "columns": {
            19: "in_progress",
            20: "statement_preparation",
            21: "at_consultant",
            22: "at_headquarters",
            23: "at_finance"
        },
        "total_column": 24
    },
    {
        "activity": "test_4_b",
        "columns": {
            25: "in_progress",
            26: "statement_preparation",
            27: "at_consultant",
            28: "at_headquarters",
            29: "at_finance"
        },
        "total_column": 30
    },
    {
        "activity": "test_6",
        "columns": {
            31: "in_progress",
            32: "statement_preparation",
            33: "at_consultant",
            34: "at_headquarters",
            35: "at_finance"
        },
        "total_column": 36
    },
    {
        "activity": "test_7",
        "columns": {
            37: "in_progress",
            38: "statement_preparation",
            39: "at_consultant",
            40: "at_headquarters",
            41: "at_finance"
        },
        "total_column": 42
    }
]
```

```python
measurement_columns = list(range(4, 49))

for column in measurement_columns:
    data[column] = pd.to_numeric(
        data[column],
        errors="coerce"
    )
```

```python
print(
    "Missing measurement values:",
    data[measurement_columns].isna().sum().sum()
)
```

```python
validation_results = []

for group in activity_groups:
    status_columns = list(group["columns"].keys())
    total_column = group["total_column"]

    calculated_total = data[status_columns].sum(axis=1)
    reported_total = data[total_column]

    invalid_rows = calculated_total != reported_total

    validation_results.append({
        "activity": group["activity"],
        "invalid_rows": int(invalid_rows.sum())
    })

validation_df = pd.DataFrame(validation_results)

validation_df
```

```python
for group in activity_groups:
    status_columns = list(group["columns"].keys())
    total_column = group["total_column"]

    calculated_total = data[status_columns].sum(axis=1)
    invalid_rows = calculated_total != data[total_column]

    if invalid_rows.any():
        print("\nActivity:", group["activity"])

        display(
            data.loc[
                invalid_rows,
                ["city_name", "year", "month", total_column]
            ]
        )
```

```python
long_records = []

for _, row in data.iterrows():

    for group in activity_groups:

        activity_name = group["activity"]

        for column_number, status_name in group["columns"].items():

            long_records.append({
                "city_name": row["city_name"],
                "city_code": row["city_code"],
                "year": row["year"],
                "month": row["month"],
                "activity": activity_name,
                "status": status_name,
                "work_order_count": row[column_number]
            })

clean_df = pd.DataFrame(long_records)
```

```python
clean_df.head(30)
```

```python
clean_df = clean_df.astype({
    "city_name": "string",
    "city_code": "int64",
    "year": "int64",
    "month": "int64",
    "activity": "string",
    "status": "string",
    "work_order_count": "int64"
})
```

```python
clean_df.info()
```

```python
print("Clean data shape:", clean_df.shape)
```

```python
duplicate_mask = clean_df.duplicated(
    subset=[
        "city_code",
        "year",
        "month",
        "activity",
        "status"
    ],
    keep=False
)

print("Duplicate rows:", duplicate_mask.sum())
```

```python
print(
    "Missing values:",
    clean_df.isna().sum().sum()
)

print(
    "Negative counts:",
    (clean_df["work_order_count"] < 0).sum()
)

print(
    "Invalid months:",
    (~clean_df["month"].between(1, 12)).sum()
)
```

```python
calculated_totals = (
    clean_df
    .groupby(
        ["city_code", "year", "month"],
        as_index=False
    )["work_order_count"]
    .sum()
    .rename(
        columns={
            "work_order_count": "calculated_total"
        }
    )
)
reported_totals = (
    data[
        ["city_code", "year", "month", 48]
    ]
    .rename(
        columns={
            48: "reported_total"
        }
    )
)
```

```python
total_check = calculated_totals.merge(
    reported_totals,
    on=["city_code", "year", "month"],
    how="left"
)

total_check["difference"] = (
    total_check["calculated_total"]
    - total_check["reported_total"]
)

total_check.head()
```

```python
print(
    "Rows with total mismatch:",
    (total_check["difference"] != 0).sum()
)
```
